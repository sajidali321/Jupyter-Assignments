```python
# Solve Question 1 complete
# Section 1: Employee Class 8 attributes define
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, experience):
        # Values save in Object 
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.experience = experience

    # Section 2: Create 3 Methods 
    
    # Method 1: Employee detail print 
    def describe_employee(self):
        print(f"--- Employee Detail: {self.name} ---")
        print(f"Phone Number: {self.number}")
        print(f"Address: {self.address}")
        print(f"Salary: Rs. {self.salary}")
        print(f"Designation/Post: {self.post}")
        print(f"Year of Joining: {self.year_of_joining}")
        print(f"Education: {self.education}")
        print(f"Experience: {self.experience} years")
        print("-" * 35)

    # Method 2: Experience update 
    def update_year_of_experience(self, new_experience):
        self.experience = new_experience
        print(f"Update: '{self.name}' experience update ho kar {self.experience} saal ho gaya hai.")

    # Method 3: Salary increase
    def increase_salary(self, increment_amount):
        self.salary += increment_amount
        print(f"Update: '{self.name}' salary mein Rs. {increment_amount} ka izafa hua. Nayi Salary: Rs. {self.salary}")


# Secion 3: create employee101 
print("========== Section 3: EMPLOYEE 101 ==========\n")
# create 8 attributes values 
employee101 = Employee("Sajid Ali", "0300-1234567", "Karachi", 50000, "Data Analyst", 2024, "Bachelors", 2)

# 4 attributes print in cell
print("--- 4 Attributes Individually ---")
print("Name:", employee101.name)
print("Post:", employee101.post)
print("Salary:", employee101.salary)
print("City:", employee101.address)
print("\n")

# call three methods
employee101.describe_employee()
employee101.increase_salary(10000) # Salary increase 10,000 
employee101.update_year_of_experience(3) # Experience 3 years
print("\n")


# Section 4: create 10 diffrent instances (employees) 
print("========== Section 4: 10 NEW EMPLOYEES ==========\n")

# 10 employees in list
employees_list = [
    Employee("Ali", "111", "Karachi", 30000, "Clerk", 2020, "B.Com", 4),
    Employee("Bilal", "222", "Lahore", 45000, "Accountant", 2021, "BBA", 3),
    Employee("Zain", "333", "Islamabad", 60000, "Manager", 2019, "MBA", 5),
    Employee("Raza", "444", "Multan", 35000, "Salesman", 2022, "Inter", 2),
    Employee("Kamran", "555", "Karachi", 80000, "Developer", 2018, "BSCS", 6),
    Employee("Faisal", "666", "Peshawar", 25000, "Guard", 2023, "Matric", 1),
    Employee("Tariq", "777", "Quetta", 55000, "HR", 2021, "BBA", 3),
    Employee("Usman", "888", "Karachi", 90000, "Team Lead", 2017, "MS", 7),
    Employee("Noman", "999", "Faisalabad", 40000, "Supervisor", 2020, "BA", 4),
    Employee("Waqas", "000", "Sukkur", 28000, "Driver", 2022, "Middle", 2)
]

# for Loop describe_employee method calling
for emp in employees_list:
    emp.describe_employee()


# Testing any 3 updations 
print("--- 3 Updations ---")
employees_list[0].increase_salary(5000) # Ali salary increased 
employees_list[2].increase_salary(3000) # Zain salary increased
employees_list[4].update_year_of_experience(8) # Kamran experience update
```
