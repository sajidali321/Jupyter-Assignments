```python
# Solve Q.1 

# Loop tab tak chalega jab tak break nahi hota
while True:
    # User se input lena
    guess = input("Guess the word: ")
    
    # Check karna ke guess sahi hai ya nahi
    if guess.lower() == "python":
        print("🎉 You guessed it!")
        break  # Sahi guess par loop ko rokne ke liye
```

```python
# Solve Q.2

users = ["Ali", "Sara", "John", "Zara", "Mike"]

for index, user in enumerate(users):
    if index == 3:
        print("Limit reached. Stop sending emails.")
        break
    print(f"Email sent to {user}")
```

```python
# Solve Q.3

temps = [22, 25, 30, 45, 28]
critical = 40

for temp in temps:
    # Agar temperature critical limit (40) tak pahonch jaye
    if temp >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break  # Loop ko break lagane k liye
        
    print(f"Temperature: {temp}")
```

```python
# Solve Q.4

cart = [100, 250, 50, 300]
total_price = 0  # price zero kia

for item in cart:
    total_price += item  # Har item ki price ko total mein jama karte jayein

# Loop khatam hone ke baad final price print karein
print(f"Total Price: {total_price}")
```

```python
# Solve Q.5
customers = [
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "John", "paid": True}
]

for customer in customers:
    # Agar customer ne bill pay nahi kiya (paid == False)
    if customer["paid"] == False:
        continue  # skip customer phir aglay customer per jana
    
    # Sirf paid customers ke liye yeh line chalegi
    print(f"Access granted to: {customer['name']}")
```

```python
# Solve Q.6

users = ["Ali", "Sara", "John"]

for user in users:
    print(f"Hello {user}, your subscription is active!")
```

```python
# Solve Q.7

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domains = []  # Domain names ko save karne ke liye khaali list

for email in emails:
    # Email ko '@' se split karke us ka doosra hissa [1] uthana
    domain = email.split("@")[1]
    
    # Domain ko list mein add karna
    domains.append(domain)

# Final list ko print karna
print(f"domains: {domains}")
```

```python
# Solve Q.8

tasks = [
    {"task": "Pay bills", "status": "done"},
    {"task": "Submit report", "status": "pending"},
    {"task": "Clean house", "status": "pending"}
]

for task_item in tasks:
    # Check karna ke status pending hai ya nahi
    if task_item["status"] == "pending":
        print(f"Overdue: {task_item['task']}")
```

```python
# Solve Q.9

while True:
    # User se number input lena aur int mein convert karna
    num = int(input("Enter a number (negative to stop): "))
    
    # Check karna k number negative hai ya nahi
    if num < 0:
        print("Negative number entered. Loop stopped!")
        break  # break karna
        
    # Agar number positive hai toh print karein
    print(f"You entered: {num}")
```

```python
# Solve Q.10

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

for day in days:
    # Check karna ke kya din weekend (Saturday/Sunday) hai
    if day == "Saturday" or day == "Sunday":
        continue  # Is din ko skip karo aur agle din par jao
        
    # Sirf workdays ke liye yeh line chalegi
    print(f"Workday: {day}")
```
