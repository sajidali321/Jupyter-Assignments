```python
def sql_query(q):
    return pd.read_sql_query(q, conn)
```

```python
# Solve Question 1

q1 = """ Select Jobtitle, AVG(TotalPaybenefits)
        as AVG_Total_PAY_BENEFITS from salaries
    where JobTitle not in ('Not Provided', 'Not provided')
    group by JobTitle

"""

sql_query(q1)
```

```python
# Solve Question 2

q2 = """ SELECT JobTitle, OvertimePay FROM Salaries
            WHERE OvertimePay > 50000
            AND OvertimePay != 'Not Provided'
            
"""

sql_query(q2)
```

```python
# Solve Question 3

q3 = """ SELECT COUNT(DISTINCT EmployeeName) AS Total_Unique_Employees FROM Salaries
            WHERE EmployeeName NOT IN ('Not Provided', 'Not provided')
"""

sql_query(q3)
```

```python
# Solve Question 4

q4 = """ SELECT EmployeeName, TotalPayBenefits FROM Salaries
            ORDER BY TotalPayBenefits DESC LIMIT 10
            
"""

sql_query(q4)
```

```python
# Solve Question 4 Other Example with complete Database

q4 = """ SELECT * FROM Salaries
            ORDER BY TotalPayBenefits DESC LIMIT 10
            
"""

sql_query(q4)
```
