### Q4. You are managing a list of students enrolled in two different courses. You need to perform various set operations to understand the student enrollment. Task: Create two sets: one for students in "Course A" and one for students in "Course B". Find the students who are in "Course A" but not in "Course B" (difference). Find students who are only in one of the two courses (symmetric difference). course_a = {"John", "Alice", "Bob", "David"} course_b = {"Alice", "Eve", "Charlie", "David"}

Using set methods

• Find students who are in Course A but not in Course B (difference)

• Find students who are only in one of the two courses (symmetric difference)

Expected Output:

Students in Course A but not in Course B: {'Bob', 'John'}

Students only in one of the two courses: {'Eve', 'John', 'Bob', 'Charlie'}

```python
# Ans Q.4
# Sets create karte hain
course_a = {"John", "Alice", "Bob", "David"}
course_b = {"Alice", "Eve", "Charlie", "David"}
```

```python
# 1. Course A but not in Course B (Difference)
diff_students = course_a.difference(course_b)
```

```python
# 2. Students only in one of the two courses (Symmetric Difference)
sym_diff_students = course_a.symmetric_difference(course_b)
```

```python
# Output display karte hain
print("Students in Course A but not in Course B:", diff_students)
print("Students only in one of the two courses:", sym_diff_students)
```

### Q.5 numbers = (10, 20, 30, 40, 50, 60, 70)
Display the first 3 elements. Display the last 3 elements. Display alternate elements.

```python
# Create kia numbers ko
numbers = (10, 20, 30, 40, 50, 60, 70)
```

```python
# 1. Pehle 3 elements display karna (Index 0 se 3 tak, par 3 shamil nahi hoga)
first_three = numbers[:3]
print("First 3 elements:", first_three)
```

```python
# 2. Aakhri 3 elements display karna (Minus index use karenge piche se start karne ke liye)
last_three = numbers[-3:]
print("Last 3 elements:", last_three)
```

```python
# 3. Alternate (ek chor kar ek) elements display karna (Step size 2 rakh kar)
alternate_elements = numbers[::2]
print("Alternate elements:", alternate_elements)
```

#### 3. Alternate walay nay kese kaam kiaa ?
#### [::2]: Jab hum do bar colon lagate hain aur aakhir mein 2 likhte hain, to iska matlab hota hai Step 2 (yani har doosra element uthao). 
#### To isne 10 uthaya, 20 ko chora, 30 uthaya, 40 ko chora, 50 uthaya, 60 ko chora, 70 uthaya, is terha agay berha.

```python
# Final Result
print("First 3 elements:", first_three)
print("Last 3 elements:", last_three)
print("Alternate elements:", alternate_elements)
```

### Q10. Set Comparison A = {1, 2, 3, 4, 5} B = {4, 5, 6, 7}

Find common elements. Find elements only in Set A. Find elements only in Set B.

```python
# Create Sets 
A = {1, 2, 3, 4, 5} 
B = {4, 5, 6, 7}
```

```python
# 1. Common elements nikalna (Intersection)
common = A.intersection(B)
print("common_numbers :" , common)
```

```python
# 2. Find elements only Set A
only_A = A.difference(B)
print("only_A :" , only_A)
```

```python
# 3. Find elements only in Set B.
only_B = B.difference(A)
print("only_B :" , only_B)
```

```python
print("common_numbers :" , common)
print("only_A :" , only_A)
print("only_B :" , only_B)
```
