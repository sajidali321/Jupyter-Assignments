#### Q.1 Find record if Sales greater than 500 and Quantity greater than 5?

```python
# Step 1- Import pandas library
import pandas as pd
```

```python
# Step-2 import csv in Data Frame
df = pd.read_csv('retail supermarket1.csv')
df
```

```python
# Step 3: records filter 
result = df[(df['Sales'] > 500) & (df['Quantity'] > 5)]
```

```python
# Step 4: Output display 
result
```

#### Q.2 Find records if Discount is more than 0.5 (50%) or Profit is negative.

```python
# Step 1: Discount > 0.5 or Profit < 0 
result_q2 = df[(df['Discount'] > 0.5) | (df['Profit'] < 0)]

# Step 2: Output display
result_q2
```

#### Q.3 Display Ship Mode and Profit columns if region is East using loc

```python
# Step-1: "Filter the East region using loc and display only two columns."
result_q3 = df.loc[df['Region'] == 'East', ['Ship Mode', 'Profit']]

# Step-2 Output display
result_q3
```

#### Q.4 How many records belong to the "Corporate" segment and the "Central" region?

```python
# Step 1: Corporate data filter 
filtered_data = df[(df['Segment'] == 'Corporate') & (df['Region'] == 'Central')]

# Step 2: Print Counting Total Records
print(f"Total Records Found: {len(filtered_data)}")

# Step 3: Display Data Filter
filtered_data
```

#### Q.5  Find all orders in the "Furniture" category but not in "Chairs" sub-category.

```python
# Step 1: Furniture category filter & Chairs sub-category exclude 
furniture_Chairs = df[(df['Category'] == 'Furniture') & (df['Sub-Category'] != 'Chairs')]

# Step 2: Filtered table display 
furniture_Chairs
```

#### Q.6 Which orders have discount (>0.3) and still earned profit (>0)?

```python
# Step 1: Discount & profit filter karein
discount_profit = df[(df['Discount'] > 0.3) & (df['Profit'] > 0)]

# Step 2: Display function use karein
display(discount_profit)

# Step 3: Total Counting
print(f"total_discount_profit = {len(discount_profit)} records")
```

#### Q.9 Show Home Office segment orders with Sales > 1000 and Quantity > 3.?

```python
# Step 1: filter home ofice, segment, quantity
h_s_q = df[(df['Segment'] == 'Home Office') & (df['Sales'] > 1000) & (df['Quantity'] > 3)]

# Step 2: Table display 
display(h_s_q)

# Step 3: Total count print 
print(f"Total Home Office Orders (Sales > 1000 & Qty > 3) = {len(h_s_q)}")
```

#### Q.10 10- Find orders where Profit Margin > 0.3
#### Profit Margin = (Profit/Sales)

```python
# Step 1: Create New Column with formula "Profit Margin"
df['Profit Margin'] = df['Profit'] / df['Sales']

# Step 2: New column (Profit Margin) par > 0.3 
p_m = df[df['Profit Margin'] > 0.3]

# Step 3: Table display
display(p_m)

# Step 4: Total count print
print(f"Total Orders with Profit Margin > 0.3 = {len(p_m)}")
```
