## string method "isnumeric" ki command use karna

```python
## Agar string mein ek bhi letter, space, ya symbol aa gaya, toh yeh False de dega:

id_card = "42101-1234567-1"  # Is mein dashes (-) hain
age = "25 years"            # Is mein space aur letters hain
```

```python
user_input = "180"

if user_input.isnumeric():
    # Agar sirf number hai toh integer mein badal dein
    age = int(user_input)
    print("Valid Age:", age)
else:
    print("Galti! Please sirf numbers likhein.")
```

## string method "find and rfind" ki command kese use karein

```python
# Ek text banate hain jahan 'Python' do baar aaya hai
text = "I love Python because Python is powerful"

# 1. find() ka istemal (Left se dhoondna)
pehla_python = text.find("Python")
print("Pehle Python ka index:", pehla_python)

# 2. rfind() ka istemal (Right se dhoondna)
aakhri_python = text.rfind("Python")
print("Aakhri Python ka index:", aakhri_python)
```

```python
text = "Smart Mind"

# Kuch aisa dhoondte hain jo text mein nahi hai
print(text.find("Java"))   # Output: -1
print(text.rfind("Java"))  # Output: -1
```

```python
sentence = "Data Analytics with Python"

if sentence.find("Python") != -1:
    print("Haan! Is line mein Python ka zikr hai.")
else:
    print("Python nahi mila.")
```
