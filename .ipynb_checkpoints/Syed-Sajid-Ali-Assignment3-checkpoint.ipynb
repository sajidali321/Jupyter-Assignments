### Q-1 # Remove the extra spaces from both ends of the string " Hello World ". (strip) ka istemal

```python
# String jismein extra spaces hain
my_string = "  Hello World  "
```

```python
# .strip() method ka istemal
clean_string = my_string.strip()
```

```python
# Output dekhne ke liye
clean_string
```

### Q-2 Remove spaces only from the left side of " Start now". (lstrip) ka istemal

```python
# String jismein shuruat mein space hai
my_string = " Start now"
```

```python
# .lstrip() method ka istemal sirf left side ka space hatane ke liye
clean_string = my_string.lstrip()
```

```python
# Output dekhne ke liye
clean_string
```

### Q-3 # Remove trailing spaces from "Finish line     ". (rstrip) ka istemal

```python
# String jismein aakhir mein space hai
my_string = "Finish line     "
```

```python
# .rstrip() method ka istemal trailing spaces hatane ke liye
clean_string = my_string.rstrip()
```

```python
clean_string
```

### Q-4 # Replace "apple" with "banana" in the sentence "I have an apple".

```python
# Asli sentence
sentence = "I have an apple"
```

```python
# .replace("purana_word", "naya_word") ka istemal
new_sentence = sentence.replace("apple", "banana")
```

```python
new_sentence
```
