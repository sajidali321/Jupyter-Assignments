## Solve Question 1

```python
# Question 1 Solution
expression_result = 30 + 4 * 3 - 6
```

```python
# Step 1: multiply (4 * 3)
step1 = 4 * 3
```

```python
print("Step 1 (Multiplication): 4 * 3 =", step1)
```

```python
# Step 2: Step 1 aur Step 2 ko add karein (30 + 12)
step2 = 30 + step1
```

```python
print("Step 2 (Addition): 30 + 12 =", step2)
```

```python
# Step 3: End mein 6 subtract kar dein (42 - 6)
final_output = step2 - 6
```

```python
print("Step 3 (Subtraction): 42 - 6 =", final_output)
```

```python
print("The output of the expression is:", expression_result)
```

## Solve Question 2

```python
# Step 1: Pehle 100 ko 4 se divide karein
step1 = 100 / 4
```

```python
print("Step 1 (Division): 100 / 4 =", step1)
```

```python
# Step 2: Phir division ke result mein 7 add karein
step2 = step1 + 7
```

```python
# Step 3: End mein 5 subtract kar dein
final_output = step2 - 5
```

```python
print("Step 3 (Subtraction): 32.0 - 5 =", final_output)
```

```python
print("Final Output :", final_output)
```

```python
# Step 2: Display "Ali" 
# Index [0] yani pehli list, aur uske andar ka Index [0] yani "Ali"
ali_name = students[0][0]
```

```python
print("1st Name :", ali_name)
```

## Solve Question 6

```python
# Step 1: Creation List
numbers = [10, 20, 30, 40, 50, 60, 70]
```

```python
print("Complete list:", numbers)
```

```python
# Step 2: Display 1st three elements 
# Start 0 to 3 par stop 
first_three = numbers[0:3]
```

```python
print("First 3 elements:", first_three)
```

```python
# Step 3: Display last three elements
# Start -3 to last
last_three = numbers[-3:]
```

```python
print("Last 3 elements:", last_three)
```
