# Assigment 9 Student Data Upgrades

```python
import csv
```

```python
# Create Grade Function
def get_grade(p):
    return 'A' if p >= 80 else 'B' if p >= 70 else 'C' if p >= 60 else 'D' if p >= 50 else 'F'

# File open Read and Write mode
with open("Student_CSV.csv", "r") as read1, open("Student_Grades_Output.csv", "w", newline="") as write1:
    reader, writer = csv.reader(read1), csv.writer(write1)
    
    # Header Grade add and write
    header = next(reader)
    writer.writerow(header + ['Student_Grade'])
    pct_idx = header.index('Percentage')
    
    # Loop
    for row in reader:
        try:
            p = float(row[pct_idx]) # Percentage convert into numbers
        except ValueError:
            p = 0.0 
            
        # Old data and new grade writing
        writer.writerow(row + [get_grade(p)])

print("Grades successfully calculate now! file save!")
```
