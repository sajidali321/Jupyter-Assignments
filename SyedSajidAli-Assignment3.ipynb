### Solve Question 5

```python
# Initial string
text = 'hello python world'

# 1. Spaces ko underscores (_) se replace kiya
replaced_text = text.replace(' ', '_')

# 2. Modified string ko underscore ke basis par split kiya (is se list banegi)
split_list = replaced_text.split('_')
```

```python
# 3. List ke elements ko hyphen (-) ke sath join kiya
joined_text = '-'.join(split_list)
```

```python
# Teeno results ko print kiya
print("1. Replaced String:", replaced_text)
print("2. Split List:     ", split_list)
print("3. Joined String:  ", joined_text)
```

### Solve Question 7

```python
# Initial product code
product_code = "SKU-12345-XYZ"
```

```python
# 1. Pehle dash (-) ki position dhoondna
# (Python mein counting 0 se shuru hoti hai, toh S=0, K=1, U=2, -=3)
first_dash_index = product_code.find("-")
```

```python
# 2. Total dashes (-) ki tadaad count karna
dash_count = product_code.count("-")
```

```python
# 1. Pehle "SKU-" prefix remove kiya
# 2. Phir "-XYZ" suffix remove kiya
numeric_part = product_code.removeprefix("SKU-")
numeric_part = numeric_part.removesuffix("-XYZ")
```

# Teeno results ko print karna
print("1. First dash position (index):", first_dash_index)
print("2. Total number of dashes:     ", dash_count)
print("3. Extracted numeric part:     ", numeric_part)

### Solve Question 9

```python
print("3. Replaced Text:   ", replaced_text)
```

```python
print("4. 'Python' Position:", python_position)
```

### Solve Question 10

```python
# Initial list
data = [12, 45, 2, 67, 34, 89, 23, 90]
```

```python
# 1. List ko ascending order (chote se bade) mein sort kiya
data.sort()
# Is step ke baad list ban gayi: [2, 12, 23, 34, 45, 67, 89, 90]
```

```python
# 2. Highest value ko remove kiya
# Kyunki list sorted hai, toh highest value hamesha aakhri index par hogi. 
# Hum .pop() use karke aakhri element hata sakte hain.
data.pop()
# Is step ke baad list ban gayi: [2, 12, 23, 34, 45, 67, 89]
```

```python
# 3. Value 50 ko index 3 par insert kiya
data.insert(3, 50)
# Is step ke baad list ban gayi: [2, 12, 23, 50, 34, 45, 67, 89]
```

```python
# 4. Finally, poorie updated list ko reverse (ulta) kiya
data.reverse()
```

```python
# Result display karein
print("Final Updated List:", data)
```
