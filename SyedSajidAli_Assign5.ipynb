```python
# Solve Question Number 1

# User se age input lene ke liye
age = int(input("Apni age enter karein: "))

# Age group classification
if age >= 0 and age <= 12:
    print("Child")
elif age >= 13 and age <= 19:
    print("Teenager")
elif age >= 20 and age <= 64:
    print("Adult")
elif age >= 65:
    print("Senior")
else:
    print("Invalid age entered!") # Agar koi minus mein age likh de
```

```python
# Solve Question Number 2

# User se teen numbers input lene ke liye
num1 = float(input("Pehla number enter karein: "))
num2 = float(input("Doosra number enter karein: "))
num3 = float(input("Teesra number enter karein: "))

# 1. Equality check karne ke liye (Kya barabar hain?)
if num1 == num2 == num3:
    print("Sari values equal hain.")
elif num1 == num2 or num2 == num3 or num1 == num3:
    print("Koi do numbers equal hain.")
else:
    print("Sare numbers mukhtalif (different) hain.")

# 2. Largest number dhoondne ke liye
if num1 >= num2 and num1 >= num3:
    largest = num1
elif num2 >= num1 and num2 >= num3:
    largest = num2
else:
    largest = num3

print(f"In mein se sab se bada number {largest} hay.")
```

```python
# Solve Question Number 3

# Pehle tuple define karein jo question mein di gayi hay
my_tuple = (5, 10, 15, 20, 25)

# User se integer input lein
user = int(input("Koi number enter karein: "))

# 'in' operator ke zariye check karein ke value tuple mein hay ya nahi
if user in my_tuple:
    print(f"Yes! {user} is tuple mein maujood hay.")
else:
    print(f"No! {user} is tuple mein maujood nahi hay.")
```

```python
# Solve Question Number 4

# 1. User se input lena
title = input("Book ka Title enter karein: ")
author = input("Author ka naam enter karein: ")
year = int(input("Year of publication enter karein: "))
price = float(input("Book ki Price enter karein: "))

# 2. Saari details ko aik tuple mein store karna
book_details = (title, author, year, price)

# 3. Expected format mein display karna
print("\n--- Book Details ---")
print(f"Title: {book_details[0]}")
print(f"Author: {book_details[1]}")
print(f"Year of Publication: {book_details[2]}")
print(f"Price: ${book_details[3]}")
```

```python
# Solve Question Number 5

# Offered subjects ki list
offered_subjects = ["math", "science", "art", "computer", "history"]

# User se input lena
user_subject = input("Enter the subject you are interested in: ")

# Input ko lowercase mein convert karna matching ke liye
user_subject_lower = user_subject.lower()

# Check karna aur naye format mein print karna
if user_subject_lower in offered_subjects:
    print(f"Subject: {user_subject} is available.")
else:
    print(f"Subject: {user_subject} is not available.")
```

```python
# Solve Question Number 6

# User se number input lene ke liye (float use kar rahe hain taake point wale numbers bhi chal sakein)
number = float(input("Enter a number: "))

# Conditions check karne ke liye
if number > 0:
    print("The number is Positive")
elif number < 0:
    print("The number is Negative")
else:
    print("The number is Zero")
```

```python
# Solve Question Number 7

# 1. username aur password define karna 
correct_username = "admin"
correct_password = "admin@123"

# 2. User se username aur password input lena
user_username = input("Username enter karein: ")
user_password = input("Password enter karein: ")

# 3. If-Elif-Else se charon conditions check karna
if user_username == correct_username and user_password == correct_password:
    print("Login Successful")
elif user_username != correct_username and user_password == correct_password:
    print("Invalid Username")
elif user_username == correct_username and user_password != correct_password:
    print("Invalid Password")
else:
    print("Invalid Username and Password")
```

```python
# Solve Question Number 9

# 1. User se inputs lena
age = int(input("Enter your age: "))
credit_score = int(input("Enter your credit score: "))
income = float(input("Enter your annual income: $ "))

# 2. Eligibility check karna
if age < 18:
    print("You are not eligible for the loan due to age.")
    
elif credit_score < 650:
    print("You are not eligible for the loan due to low credit score.")
    
else:
    # Age aur Base Credit Score theek hay, ab income ke dono combinations check karte hain
    condition1 = (credit_score >= 650 and income >= 30000)
    condition2 = (credit_score >= 700 and income >= 25000)
    
    if condition1 or condition2:
        print("You are eligible for the loan!")
    else:
        print("You are not eligible for the loan due to low income.")
```

```python
# Solve Question Number 10

# User se letter input lena
char = input("Enter a single letter: ")

# 1. Check karna ke input sirf ek character ho aur wo alphabet ho
if len(char) == 1 and char.isalpha():
    
    # Input ko choti ABC (lowercase) mein convert karna taake checking aasan ho
    char_lower = char.lower()
    
    # 2. Vowels check karna
    if char_lower in ['a', 'e', 'i', 'o', 'u']:
        print(f"{char} is a Vowel")
    else:
        print(f"{char} is a Consonant")

else:
    # Agar 1 se zyada characters hon ya number/symbol ho
    print("Please enter valid alphabet")
```
