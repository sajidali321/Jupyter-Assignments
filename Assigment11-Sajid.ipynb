#### Q.1 Check Duplicate Records

```python
# Step 1 read import pandas and csv file
import pandas as pd

df = pd.read_csv("cars.csv")
```

```python
df
```

```python
# Step 2: Total duplicate rows checking
duplicate_count = df.duplicated().sum()
print("Total Duplicate Rows:", duplicate_count)

# Step 3: Duplicate display 
duplicate_rows = df[df.duplicated()]
duplicate_rows
```

#### Q.2 Drop Duplicate Records

```python
# Step 1: Duplicate rows delete (drop) karna
df = df.drop_duplicates()

# Step 2: Verify rows aur columns 
print("Duplicates drop kay bad dataset shape:", df.shape)

# Step 3: again check duplicates data
print("Check duplicate rows:", df.duplicated().sum())
```

#### Q.3 Check missing or null values

```python
# Dataset missing / null values checking
df.isnull().sum()
```

#### Q.4 Drop missing or null values

```python
# Step 1: Missing/ null values rows deleting (drop)
df = df.dropna()

# Step 2: Verify null values drop or not
print("\nMissing values drop k baad dataset ka shape:", df.shape)
print("\nRemaing null values:")
print(df.isnull().sum())
```

#### Q.5 Show top 10 average MSRP (Price) per car brand (rounded figure)?

```python
# Step 1: Every car brand (Make) average MSRP calculate and round 
top_10_brands = (
    df.groupby("Make")["MSRP"]
    .mean()
    .round()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

# Step 2: Result display
top_10_brands
```

#### Q.6 Show the average MSRP (price) per years (rounded figure)?

```python
# Step 1: Every (Year) average MSRP calculate and round 
avg_msrp_per_year = df.groupby("Year")["MSRP"].mean().round().reset_index()

# Step 2: Result display 
avg_msrp_per_year
```

#### Q.7 Show average Engine HP (power) rounded figure and MPG (efficiency) rounded figure per year?
#### Hint: Average of (city mpg + highway MPG)


```python
# Step 1: Every car combined Average MPG (efficiency) 
df["MPG"] = (df["city mpg"] + df["highway MPG"]) / 2

# Step 2: Every (Year) Engine HP and MPG average nikalna aur round karna
avg_hp_mpg_year = (
    df.groupby("Year")[["Engine HP", "MPG"]].mean().round().reset_index()
)

# Step 3: Result table display
avg_hp_mpg_year
```

#### Q.8 What percentage of cars are Compact, Midsize, or Large?¶

```python
# Step 1: Vehicle percentage 
vehicle_size_pct = (df["Vehicle Size"].value_counts(normalize=True) * 100).round(2).reset_index()

# Step 2: Columns Names Set
vehicle_size_pct.columns = ["Vehicle Size", "Percentage (%)"]

# Step 3: Result display 
vehicle_size_pct
```

#### Q.9 Which vehicle styles are the most sold (most common) in the dataset?

```python
# Step 1: Vehicle Style column count set
vehicle_styles = df["Vehicle Style"].value_counts().reset_index()

# Step 2: Columns headings 
vehicle_styles.columns = ["Vehicle Style", "Count"]

# Step 3: Top most common vehicle styles display
vehicle_styles.head(10)
```

#### Q.10 Which car brands have the most vehicles priced at $40,000 or more and with a popularity score of at least 1000?

```python
# Step 1: Filter condition (MSRP >= 40,000 aur Popularity >= 1000)
filtered_cars = df[(df["MSRP"] >= 40000) & (df["Popularity"] >= 1000)]

# Step 2: Car brands (Make) count and table formating
popular_expensive_brands = (
    filtered_cars["Make"].value_counts().reset_index()
)
popular_expensive_brands.columns = ["Make", "Count"]

# Step 3: Top brands display 
popular_expensive_brands.head(10)
```
