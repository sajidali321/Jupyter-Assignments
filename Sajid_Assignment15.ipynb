```python
import pandas as pd
import sqlite3 as sql
```

```python
conn = sql.connect('factbook.db')
conn
```

```python
def sql_query(query):
    return pd.read_sql_query(query, conn)
```

```python
# Q.1 List all cities where the area is greater than 5000, along with their population.
```

```python
q = '''
    select cn.name city_name, cn.population city_population, ca.area country_area
    from cities cn
    inner join facts ca
    on cn.facts_id = ca.id
    where ca.area > 5000
'''
sql_query(q)
```

```python
# Q.2 List the names of capital cities where the death rate is greater than 2% and the birth rate is less than 20%
```

```python
q = '''
    select cc.name capital_city, db.death_rate, db.birth_rate
    from cities cc
    inner join facts db
    on cc.facts_id = db.id
    where cc.capital == 1 and db.death_rate > 2 and db.birth_rate < 20
'''
sql_query(q)
```

```python
# Q.3 List the cities with a population growth rate between 1% and 5%, sorted in desc by population growth.
```

```python
q = '''
    select cn.name city_name, pg.population_growth
    from cities cn
    inner join facts pg
    on cn.facts_id = pg.id
    where pg.population_growth between 1 and 5
    order by pg.population_growth desc
'''
sql_query(q)
```

```python
# Q.4 Find countries that start with 'A' and do not have any cities listed in the cities table. 
# The query should return the country name and city name. Using left join
```

```python
q = '''
    select cty.name country_name, cn.name city_name
    from facts cty
    left join cities cn
    on cty.id = cn.facts_id
    where cty.name like 'A%' and cn.name is null
'''
sql_query(q)
```

```python
# Q.5 List all capital cities and their corresponding death rate using a LEFT JOIN.
```

```python
q = '''
    select cc.name capital_city, dr.death_rate
    from cities cc
    left join facts dr
    on cc.facts_id = dr.id
    where cc.capital == 1
'''
sql_query(q)
```
