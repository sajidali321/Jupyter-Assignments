### Q-1 # Remove the extra spaces from both ends of the string " Hello World ". (strip) ka istemal

```python
# String jismein extra spaces hain
my_string = "  Hello World  "
```

```python
# .strip() method ka istemal
clean_string = my_string.strip()
```

```python
# Output dekhne ke liye
clean_string
```

### Q-2 Remove spaces only from the left side of " Start now". (lstrip) ka istemal

```python
# String jismein shuruat mein space hai
my_string = " Start now"
```

```python
# .lstrip() method ka istemal sirf left side ka space hatane ke liye
clean_string = my_string.lstrip()
```

```python
# Output dekhne ke liye
clean_string
```

### Q-3 # Remove trailing spaces from "Finish line     ". (rstrip) ka istemal

```python
# String jismein aakhir mein space hai
my_string = "Finish line     "
```

```python
# .rstrip() method ka istemal trailing spaces hatane ke liye
clean_string = my_string.rstrip()
```

```python
clean_string
```

### Q-4 # Replace "apple" with "banana" in the sentence "I have an apple". (replace) ka istemal

```python
# Asli sentence
sentence = "I have an apple"
```

```python
# .replace("purana_word", "naya_word") ka istemal
new_sentence = sentence.replace("apple", "banana")
```

```python
new_sentence
```

### Q-5 # Remove the .jpg suffix from "picture.jpg". (removesuffix) ka istemal

```python
# Method -1
# File ka naam suffix ke saath
filename = "picture.jpg"
```

```python
# .removesuffix() ka istemal
clean_jpg = filename.removesuffix(".jpg")
```

```python
clean_jpg
```

### Q-6 # Remove the "Mr. " prefix from "Mr. John". (removeprefix) ka istemal

```python
# Naam prefix ke saath
name = "Mr. John"
```

```python
# .removeprefix() ka istemal prefix hatane ke liye
clean_name = name.removeprefix("Mr. ")
```

```python
clean_name
```

### Q-7 # How many times does the letter "e" appear in "experience"? (count) ka istemal

```python
# String jismein count karna hai
word = "experience"
```

```python
# .count() method ka istemal 'e' ko ginne ke liye
count_e = word.count('e')
```

```python
count_e
```
